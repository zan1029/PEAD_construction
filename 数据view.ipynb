```python
import pyarrow.parquet as pq
import glob, os

files = sorted(glob.glob("build/*.parquet"))
for f in files:
    pf = pq.ParquetFile(f)
    print("="*80)
    print(os.path.basename(f))
    print("rows:", pf.metadata.num_rows, "| row_groups:", pf.num_row_groups)
    print("schema:", pf.schema_arrow)

    head_df = pf.read_row_group(0).to_pandas()
    print("head:")
    print(head_df.head())

    tail_df = pf.read_row_group(pf.num_row_groups - 1).to_pandas()
    print("tail:")
    print(tail_df.tail())
```
